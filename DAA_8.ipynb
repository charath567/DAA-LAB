```python
from collections import deque

class Graph:
    def __init__(self, directed=False):
        self.graph = {}
        self.directed = directed

    def add_edge(self, u, v):
        """Add an edge between vertex u and vertex v."""
        if u not in self.graph:
            self.graph[u] = []
        if v not in self.graph:
            self.graph[v] = []

        self.graph[u].append(v)
        if not self.directed:
            self.graph[v].append(u)

    def dfs(self, start_node, visited=None):
        """
        Depth-First Search (DFS) Traversal - Recursive
        Explores as deep as possible along each branch before backtracking.
        """
        if visited is None:
            visited = set()

        visited.add(start_node)
        print(start_node, end=" ")

        for neighbor in self.graph.get(start_node, []):
            if neighbor not in visited:
                self.dfs(neighbor, visited)

    def bfs(self, start_node):
        """
        Breadth-First Search (BFS) Traversal - Iterative
        Explores neighbor nodes layer-by-layer before moving deeper.
        """
        visited = set([start_node])
        queue = deque([start_node])

        while queue:
            node = queue.popleft()
            print(node, end=" ")

            for neighbor in self.graph.get(node, []):
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)


# Build Graph
g = Graph(directed=False)
g.add_edge(0, 1)
g.add_edge(0, 2)
g.add_edge(1, 3)
g.add_edge(1, 4)
g.add_edge(2, 5)

# Interactive User Input
try:
    start_node = int(input("Enter starting node for traversal (e.g., 0): "))

    if start_node not in g.graph:
        print(f"Node {start_node} does not exist in the graph.")
    else:
        print(f"\nDFS Traversal starting from node {start_node}:")
        g.dfs(start_node)
        print("\n")

        print(f"BFS Traversal starting from node {start_node}:")
        g.bfs(start_node)
        print()

except ValueError:
    print("\nError: Please enter a valid integer for the starting node!")
```
